# Option 1 - Document System

The document system should be able to handle the creation of different document types where each type has a different creation method through a common interface.

Using Python's library for abstract classes and methods where it is commonly used to create interfaces. "abc" refers to a base class when inherited from creates an abstract class.

```python
from abc import ABC, abstractmethod
```

The Document abstract class contains the create, open and save abstract methods.

```python
# Factory product interface
class Document(ABC):
    @abstractmethod
    def create(self):
        """Abstract method to create a document"""
        pass

    @abstractmethod
    def open(self):
        """Abstract method to open a document"""
        pass

    @abstractmethod
    def save(self):
        """Abstract method to save a document"""
        pass
```

Next, the concrete document classes for each document type. Each type has seperate logic for their respective operations.

```python
# Concrete Factory Products
class PDFDocument(Document):
    def create(self):
        """Abstract method to create a PDF document"""
        print("Creating PDF document")

    def open(self):
        """Abstract method to open a PDF document"""
        print("Opening PDF document")

    def save(self):
        """Abstract method to save a PDF document"""
        print("Saving PDF document")


class CSVDocument(Document):
    def create(self):
        """Abstract method to create a CSV document"""
        print("Creating CSV document")

    def open(self):
        """Abstract method to open a CSV document"""
        print("Opening CSV document")

    def save(self):
        """Abstract method to save a CSV document"""
        print("Saving CSV document")


class HTMLDocument(Document):
    def create(self):
        """Abstract method to create a HTML document"""
        print("Creating HTML document")

    def open(self):
        """Abstract method to open a HTML document"""
        print("Opening HTML document")

    def save(self):
        """Abstract method to save a HTML document"""
        print("Saving HTML document")
```

The Concrete interface of the DocumentCreator class defines the Factory Method and how the documents are created.

```python
# Factory Creator Interface
class DocumentCreator(ABC):
    @abstractmethod
    def factory_method(self):
        """Define the factory method to create a document"""
        pass

    def create_document(self):
        """Abstract method to create a document using the factory method"""
        document = self.factory_method()
        document.create()
        document.open()
        document.save()
```

The concrete class creators will return the correct type needed from the client which alters the factory_method from the base DocumentCreator class depending on which type is chosen.

```python
# Concrete Factory Creators
class PDFDocumentCreator(DocumentCreator):
    def factory_method(self):
        """Define the factory method to create a PDF document"""
        print("Starting PDF document creator")
        return PDFDocument()


class CSVDocumentCreator(DocumentCreator):
    def factory_method(self):
        """Define the factory method to create a CSV document"""
        print("Starting CSV document creator")
        return CSVDocument()


class HTMLDocumentCreator(DocumentCreator):
    def factory_method(self):
        """Define the factory method to create a HTML document"""
        print("Starting HTML document creator")
        return HTMLDocument()
```

# Test Runs

This test run simulates the client code creating a document and through a single interface is able to choose which type is needed and the different methods performed on the document.

```python
def test_code(creator: DocumentCreator):
    """Abstract test code that tests the document creation process using each concrete factory creator"""
    creator.create_document()

types = [
    PDFDocumentCreator(),
    CSVDocumentCreator(),
    HTMLDocumentCreator()
]

for type in types:
    test_code(type)
    print('\n')
```
