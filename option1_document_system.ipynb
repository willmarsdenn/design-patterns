# Option 1 - Document System

The document system should be able to handle the creation of different document types where each type has a different creation method through a common interface.

Using Python's library for abstract classes and methods where it is commonly used to create interfaces. "abc" refers to a base class when inherited from creates an abstract class.

```python
from abc import ABC, abstractmethod
```

```python
# Factory product interface
class Document(ABC):
    @abstractmethod
    def create(self):
        """Abstract method to create a document"""
        pass

    @abstractmethod
    def open(self):
        """Abstract method to open a document"""
        pass

    @abstractmethod
    def save(self):
        """Abstract method to save a document"""
        pass
```
