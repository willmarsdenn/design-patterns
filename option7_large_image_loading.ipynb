# Option 7 - Large Image Loading

The large image loading system should be able to manage large images efficiently by only loading when needed.

Using Python's library for abstract classes and methods where it is commonly used to create interfaces. "abc" refers to a base class when inherited from creates an abstract class.

```python
from abc import ABC, abstractmethod
```

Image is an abstract base class that has a method for displaying the image.

```python
class Image(ABC):
    @abstractmethod
    def display(self):
        pass

    def reload(self):
        pass
```

The LargeImage class will be used to to simulate loading a large, resource intensive image from local storage. It contains the logic for retrieving the image from local storage, 

```python
class LargeImage(Image):
    def __init__(self, filename):
        # Initialise the LargeImage with a filename and load the image from storage.
        self.filename = filename
        self.load_image_from_storage()

    def load_image_from_storage(self):
        print(f"Loading large image from {self.filename}")

    def display(self):
        print(f"Displaying large image: {self.filename}")

    def reload(self):
        print(f"Reloading large image: {self.filename}")
        self.load_image_from_storage()
```

The ImageProxy class will store common, shared information about the LargeImage object. An example in this solution is the filename. 

```python
class ImageProxy(Image):
    def __init__(self, filename):
        # Initialise the ImageProxy with a filename and set the large_image to None.
        self.filename = filename
        self.large_image = None

    def display(self):
        # If the large_image is None, create a new LargeImage instance and display it. Otherwise, use the cached large_image and display it.
        if self.large_image is None:
            self.large_image = LargeImage(self.filename)
        else:
            print(f"Using cached large image: {self.filename}")
        self.large_image.display()

    def reload(self):
        # If the large_image is None, create a new LargeImage instance. Otherwise, reload the existing large_image.
        if self.large_image is None:
            self.large_image = LargeImage(self.filename)
        else:
            self.large_image.reload()
```
