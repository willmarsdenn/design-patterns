# Option 8 - Support Ticket System

The support ticket system should be able to escalate incoming tickets from level 1, level 2 and level 3 if the previous level is unable to fix the issue.

Using Python's library for abstract classes and methods where it is commonly used to create interfaces. "abc" refers to a base class when inherited from creates an abstract class.

```python
from abc import ABC, abstractmethod
```

The Ticket class will hold information such as the ticket's ID, priority and description. 

```python
class Ticket:
    def __init__(self, ticket_id, description, priority):
        self.ticket_id = ticket_id
        self.description = description
        self.priority = priority
```

With the Chain of Respoonsbility pattern, the ticket handler class will be used to define methods for moving the ticket between levels and processing the ticket.

```python
# Chain of Responsibility Handler interface
class TicketHandler(ABC):
    def __init__(self):
        # Next handler in the chain is set to None initially
        self.next_handler = None

    def next(self, handler):
        # Set the next handler in the chain and return it for chaining
        self.next_handler = handler
        return handler

    def handle(self, ticket):
        # If the current handler can handle the ticket, process it; otherwise, pass it to the next handler in the chain
        if self.handle(ticket):
            self.process(ticket)
        elif self.next_handler is not None:
            # If the current handler cannot handle the ticket and there is a next handler, pass the ticket to the next handler
            print(f"{self.__class__.__name__} is unable to handle ticket number: {ticket.ticket_id}")
            self.next_handler.handle(ticket)
        else:
            # If there is no next handler and the ticket cannot be handled, print a message indicating that the ticket has been escalated beyond level 3
            print(f"{self.__class__.__name__} is unable to handle ticket number: {ticket.ticket_id}")
            print(f"Ticket {ticket.ticket_id} has been escalated beyond level 3.")

    @abstractmethod
    def handle(self, ticket):
        pass

    @abstractmethod
    def process(self, ticket):
        pass
```

Concrete Handlers will recieve the request and determine whether to handle the ticket or pass it further along the chain of responsibility. If there is no other level, a message will be displayed saying it has been escalated beyond level 3.

```python
# Concrete Handlers
class LevelOneSupport(TicketHandler):
    def handle(self, ticket):
        # Level one support handle tickets with low severity
        return ticket.severity == "low"

    def process(self, ticket):
        # Level one support processes the ticket
        print(f"Level 1 is processing the ticket number: {ticket.ticket_id}\n {ticket.description}")


class LevelTwoSupport(TicketHandler):
    def handle(self, ticket):
        # Level two support handle tickets with medium severity
        return ticket.severity == "medium"

    def process(self, ticket):
        # Level two support processes the ticket
        print(f"Level 2 is processing the ticket number: {ticket.ticket_id}\n {ticket.description}")


class LevelThreeSupport(TicketHandler):
    def handle(self, ticket):
        # Level three support handle tickets with high severity
        return ticket.severity == "high"

    def process(self, ticket):
        # Level three support processes the ticket
        print(f"Level 3 is processing the ticket number: {ticket.ticket_id}\n {ticket.description}")
```
